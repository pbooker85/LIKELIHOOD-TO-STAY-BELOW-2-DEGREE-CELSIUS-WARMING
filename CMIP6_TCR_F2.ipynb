```python
ssp_keys = ['SSP119', 'SSP126', 'SSP245', 'SSP370low', 'SSP370', 'SSP434', 'SSP460', 'SSP534', 'SSP585']
df_all_scenarios = df_all_scenarios[ssp_keys]

# Convert the DataFrame to a NumPy array for matrix calculations.
# X will have shape (n_years, n_scenarios) where n_scenarios should now be 9.
X = df_all_scenarios.values

# 1) your baseline (length 9, sums to 1)
baseline = np.array([0.03,0.08,0.26,0.10,0.08,0.20,0.10,0.12,0.03])

# 2) concentration parameters 
# N = (1-5) high variability
# N = (10-20) Medium Variability
# N = (> 50) low variability
N = 9  
alpha = baseline * N        # sum(alpha)=5  

# 3) draw 1000 weight vectors
n_reps = 1000
weights_draws = np.random.dirichlet(alpha, size=n_reps)  
# shape (1000, 9)

# 4) apply them to scenario matrix X (n_years×9)
buffer = X.dot(weights_draws.T)  
# buffer.shape == (n_years, 1000)

df_weighted = pd.DataFrame(
    buffer, 
    index=df_all_scenarios.index,
    columns=[f"Draw_{i+1}" for i in range(n_reps)]
)
```

```python
#############################################################################################
# Create a dataframe for parameters in our ode function...
#############################################################################################
df_table = pd.read_csv('Documents/Masters Thesis CSV/Geoffroy_parameters_EBM1_CMIP6.csv')
Model=df_table['Model']
ECS=df_table['ECS']
Gamma=df_table['gamma']
F2=df_table['F_2x']
Lambda=df_table['lambda']
cs=df_table['C']
cd=df_table['C0']
# Create Dataframe
df_constants=pd.DataFrame({'Model':Model,'F2':F2,'Gamma':Gamma, 'Lambda':Lambda, 'ECS':ECS, 'CS':cs, 'CD':cd})
df_constants
```

```python
# Remove rows with NaN values
df_constants_cleaned = df_constants.dropna()
# Remove duplicate models, keeping the first occurrence
df_constants_unique = df_constants.drop_duplicates(subset='Model')
# Remove rows with NaN values
df_constants_cleaned = df_constants_unique.dropna()
```

```python
# Constants
F2 = np.array(df_constants_cleaned.iloc[:,1])
gamma = np.array(df_constants_cleaned.iloc[:,2])
lamb = np.array(df_constants_cleaned.iloc[:,3])
ECS_std = np.array(df_constants_cleaned.iloc[:,4])
cs = np.array(df_constants_cleaned.iloc[:,5])
cd = np.array(df_constants_cleaned.iloc[:,6])

def co2_ratio(c, c0=277.454010):
    # Compute the normalized CO₂ concentration (c/c0)
    result = c / c0 
    return result

# Convert the entire df_weighted DataFrame to a NumPy array.
# (Assuming each column is a scenario's global CO₂ time series.)
c = np.array(df_weighted)
ratio = co2_ratio(c)

# Define the log-linear function globally.
def log_linear(ratio, force, base=2):
    """
    Compute radiative forcing from the normalized CO₂ ratio.
    
    Parameters:
        ratio (np.ndarray): Array of normalized CO₂ concentrations.
        force (float): Forcing at CO₂ doubling (F2).
        base (float): Logarithm base (default 2).
        
    Returns:
        np.ndarray: Radiative forcing time series.
    """
    return force * np.log(ratio) / np.log(base)

# Coupled ODE function for the Energy Balance Model (EBM)
def secondorder(T, t, gamma, lamb, f, cs, cd, e=1):
    """
    Two-layer EBM ODEs.
    
    Parameters:
        T: [T_atmosphere, T_deep]
        t: time variable (dummy for odeint)
        gamma: exchange parameter
        lamb: feedback parameter
        f: radiative forcing at the current time step
        cs: heat capacity of the surface layer
        cd: heat capacity of the deep ocean
        e: emissivity factor (default 1)
    
    Returns:
        list: [dT_atmosphere/dt, dT_deep/dt]
    """
    dT0dt = (f - T[0]*(lamb + gamma) + gamma*T[1] - (e - 1)) / cs    
    dT1dt = (gamma * (T[0] - T[1])) / cd
    return [dT0dt, dT1dt]

def run(time, forcing, gamma, lamb, cs, cd, label='info about the run'):
    """Run our climate emulator according to Eqn. 11."""
    print(f"start model simulation {label} ...")
    T_start = np.array([0, 0])
    Ts = np.zeros((len(time), 2))
    Ts[0, :] = T_start
    for i in range(len(time) - 1):
        ts = [time[i], time[i+1]]
        f = forcing[i]  # For current year
        T = odeint(secondorder, T_start, ts, args=(gamma, lamb, f, cs, cd))
        T_start = T[-1, :]
        Ts[i+1, :] = T_start
    print(f"Finished simulation {label}: last f = {f:.2f} W/m², TS = {Ts[i,0]:.2f} K, TD = {Ts[i,1]:.2f} K")
    return Ts

# Define the simulation time.
yr1, yr2 = 0, 2500.0
nyrs = 2501
t = np.linspace(yr1, yr2, nyrs)

# Compute the CO₂ ratio from df_weighted. We assume df_weighted has at least 9 columns.
def co2_ratio(c, c0=277.454010):
    return c / c0

# Use the first 9 columns (one per SSP scenario) from df_weighted.
c = np.array(df_weighted.iloc[:, 0:9])
ratio = co2_ratio(c)  # ratio will have shape (nyrs, 9)

# Loop over each model in df_constants_cleaned and over each forcing scenario (0-8)
for index, row in df_constants_cleaned.iterrows():
    model_name   = row['Model']
    force_value  = row['F2']
    gamma_value  = row['Gamma']
    lambda_value = row['Lambda']
    cs_value     = row['CS']
    cd_value     = row['CD']

    # For each forcing scenario (columns 0 to 8 of the ratio array)
    for iforc in range(9):
        # Compute radiative forcing using the log-linear function.
        # log_linear returns an array of shape (nyrs, 9); select column iforc.
        Force_all = log_linear(ratio, force=force_value)
        Force = Force_all[:, iforc]

        # Run the simulation for this model and forcing scenario.
        Tout = run(t, Force, gamma=gamma_value, lamb=lambda_value, cs=cs_value, cd=cd_value, 
                   label=f'{model_name}_log{iforc}')
        Temp_sur  = Tout[:, 0]
        Temp_deep = Tout[:, 1]

        # Prepare a DataFrame for output.
        df_out = pd.DataFrame({
            'Year': t,
            'F': Force,
            'TS': Temp_sur,
            'TD': Temp_deep
        })
        
        # Generate a filename based on the model name and forcing scenario.
        filename = f"{model_name}_log{iforc}.csv"
        print("Saving:", filename)

        # Save the DataFrame to a CSV file.
        df_out.to_csv(filename, index=False)
```

```python
import glob
import pandas as pd
import numpy as np

def compute_decadal_probability(model_name, threshold=2.0, start_year=2020, end_year=2100):
    """
    For a given model, read all simulation CSV files,
    extract the period from start_year to end_year,
    and compute the probability (fraction of years) that the surface
    temperature anomaly (TS) exceeds the threshold in each decade.
    
    Parameters:
        model_name (str): The model name (used in filenames, e.g., 'CESM2-WACCM')
        threshold (float): The temperature anomaly threshold (default = 2.0 K)
        start_year (int): The starting year for analysis
        end_year (int): The ending year for analysis
    
    Returns:
        pandas.Series: Indexed by decade, with the probability that TS >= threshold.
    """
    file_pattern = f"{model_name}_log*.csv"
    files = glob.glob(file_pattern)
    if not files:
        print(f"No files found for model {model_name}")
        return None
    
    scenario_list = []
    
    for f in files:
        df = pd.read_csv(f)
        # Check if 'Year' is a column; if not, assume the index represents the year.
        if 'Year' not in df.columns:
            df.reset_index(inplace=True)
            df.rename(columns={'index': 'Year'}, inplace=True)
        # Ensure the Year column is numeric.
        df['Year'] = pd.to_numeric(df['Year'], errors='coerce')
        # Filter for the desired period.
        df_period = df[(df['Year'] >= start_year) & (df['Year'] <= end_year)].copy()
        # Set the index to 'Year' and extract the 'TS' column.
        ts_series = df_period.set_index('Year')['TS']
        scenario_list.append(ts_series)
    
    # Concatenate all scenarios (columns are different scenarios).
    df_model = pd.concat(scenario_list, axis=1)
    
    # Create a new column "Decade" by flooring the Year to the nearest decade.
    df_model['Decade'] = (df_model.index // 10) * 10
    
    # For each decade, calculate the fraction (per scenario) that TS exceeds the threshold.
    decadal_fraction = df_model.groupby('Decade').apply(
        lambda x: (x.drop(columns='Decade') >= threshold).mean()
    )
    
    # Average across scenarios to get a single probability per decade.
    overall_prob = decadal_fraction.mean(axis=1)
    
    return overall_prob

# Example usage for one model:
model_name = 'CESM2-WACCM'
cesm2_prob = compute_decadal_probability(model_name, threshold=2.0, start_year=2020, end_year=2100)
print(f"Decadal probabilities for {model_name}:")
print(cesm2_prob)
```

```python
# List of 22 models (as provided)
models = [
    "AWI-CM-1-1-MR",
    "BCC-CSM2-MR",
    "BCC-ESM1",
    "CAMS-CSM1-0",
    "FGOALS-f3-L",
    "CanESM5",
    "ACCESS-CM2",
    "E3SM-1-0",
    "INM-CM4-8",
    "INM-CM5-0",
    "IPSL-CM6A-LR",
    "MIROC6",
    "MPI-ESM1-2-HR",
    "MRI-ESM2-0",
    "GISS-E2-1-G",
    "GISS-E2-1-H",
    "GISS-E2-2-G",
    "CESM2-WACCM",
    "CESM2",
    "GFDL-ESM4",
    "NESM3",
    "SAM0-UNICON"
]

# Container for results; each model's decadal probabilities will be stored here.
results = {}

# Loop through each model and compute its decadal probability series.
for model in models:
    prob_series = compute_decadal_probability(model, threshold=2.0, start_year=2020, end_year=2100)
    if prob_series is not None:
        results[model] = prob_series

# Combine the results into a single DataFrame. The index will be the decade (e.g., 2020, 2030, etc.).
df_prob = pd.DataFrame(results)
print("Decadal probabilities of TS >= 2 K for all models:")
print(df_prob)

# Optionally, save the combined results to a CSV file.
df_prob.to_csv("Decadal_Probability_TS_ge_2K_all_models.csv", index=True)

# Calculate the multimodel mean for each decade (average over the 22 models)
multimodel_mean = df_prob.mean(axis=1)

# Plot the multimodel mean as a line graph.
plt.figure(figsize=(10, 6))
plt.plot(multimodel_mean.index, multimodel_mean, marker='o', linestyle='-', color='blue', linewidth=2, label='Weighted Multi-Model Mean')
plt.xlabel("Decade")
plt.ylabel("Probability TS ≥ 2 °C")
plt.title("Weighted Multimodel Mean Probability of TS ≥ 2 °C (2020-2100)")
plt.legend(loc='upper left')
plt.grid(True)
plt.tight_layout()
plt.savefig('Multimodel Mean Probability of TS ≥ 2 °C.png')
plt.show()
```
